## 1. Walk-forward 2000–2021 and holdout 2022+

Retrained every 2 years on all earlier data (with a 21 + *h* day gap so no training target overlaps the test block), then one model trained on data before 2022 is tested on 2022 → today. The volatility-band baseline is run alongside for reference.

```python
pred = rrm.run_validation(table, feats, ["vol_band", MODEL], log=lambda *a: None)
pred[pred.model == MODEL].to_parquet(DATA_PROCESSED / f"range_pred_{MODEL}.parquet")
print(f"{(pred.model == MODEL).sum():,} out-of-sample predictions saved")
```

## 2. Scores

- **MAE %**: error of the median forecast, in % of price.
- **Pinball ×100**: quantile loss of the whole 10/50/90% band (lower is better).
- **Coverage 80%**: share of days inside the 10–90% band (ideal 0.80).
- **Skill**: 1 − MAE / MAE of the volatility band (> 0 beats it).

```python
m = rrm.metrics(pred)
m["target"] = m["target"].map({"up": "max high", "down": "min low"})
show = m.pivot_table(index=["split", "h", "target"], columns="model",
                     values=["mae_pct", "pinball_x100", "coverage_80", "mae_skill_vs_volband"])
show.style.format("{:.3f}")
```

## 3. Forecast vs. reality on the holdout

```python
h_pred = pred[(pred.model == MODEL) & (pred.split == "holdout")]
fig, axes = plt.subplots(2, 1, figsize=(14, 7), sharex=True)
for ax, h in zip(axes, (5, 21)):
    for t, col, lab in (("up", C_MAX, "max"), ("down", C_MIN, "min")):
        band_vs_actual(ax, h_pred[(h_pred.h == h) & (h_pred.target == t)], col, lab)
    ax.set_title(f"{MODEL_NAMES[MODEL]}: max rise / max fall over the next {h} days, holdout 2022+")
axes[0].legend(ncol=3, fontsize=8, loc="upper left")
plt.tight_layout(); plt.show()
```

## 4. Choosing the regularization (walk-forward only)

The first setting (depth 5, learning rate 0.05) overfit. Three settings were compared on test blocks 2000–03, 2008–11 and 2016–19 only; the holdout plays no part in this choice.

```python
from catboost import CatBoostRegressor
from sp500.range_models import pinball
cfgs = {"depth 5, lr 0.05": dict(depth=5, learning_rate=0.05, iterations=500),
        "depth 4, lr 0.03, l2 10 (chosen)": dict(depth=4, learning_rate=0.03, iterations=400, l2_leaf_reg=10),
        "depth 3, lr 0.03, l2 10, rsm 0.5": dict(depth=3, learning_rate=0.03, iterations=600, l2_leaf_reg=10, rsm=0.5)}
res = {}
for h in (5, 21):
    for y0 in (2000, 2008, 2016):
        te = np.flatnonzero((table.index.year >= y0) & (table.index.year < y0 + 4))
        tr = np.arange(te[0] - 21 - h)
        s = table[f"sigma_{h}"].to_numpy()
        for tg in ("up", "down"):
            y = table[f"{tg}_{h}"].to_numpy() / s
            ok, tt = tr[np.isfinite(y[tr])], te[np.isfinite(y[te])]
            for k, c in cfgs.items():
                mdl = CatBoostRegressor(loss_function="MultiQuantile:alpha=0.1,0.5,0.9", verbose=0, thread_count=2,
                                        random_seed=42, allow_writing_files=False, **c)
                mdl.fit(table[feats].iloc[ok], y[ok])
                q = np.sort(mdl.predict(table[feats].iloc[tt]), 1) * s[tt, None]
                res.setdefault((f"{tg} {h}d", k), []).append(pinball(y[tt] * s[tt], q) * 100)
pd.Series({k: np.mean(v) for k, v in res.items()}).unstack().style.format("{:.3f}").highlight_min(axis=1, color="#d7e8fb")
```

## 5. What the model uses

```python
from sp500.range_models import MODELS
imps = []
for h in (5, 21):
    for t in ("up", "down"):
        y = table[f"{t}_{h}"] / table[f"sigma_{h}"]
        ok = y.notna() & table[f"sigma_{h}"].notna()
        mdl = MODELS["catboost"]().fit(table.loc[ok, feats], y[ok])
        imp = mdl.importance(feats)
        imps.append(imp / imp.sum())
imp = pd.concat(imps, axis=1).mean(axis=1).sort_values(ascending=False)
top = imp.head(20)[::-1]
fig, ax = plt.subplots(figsize=(9, 6))
ax.barh(top.index, top.values * 100, color="#257de0")
ax.set_xlabel("share of importance (%)"); ax.set_title("catboost: top 20 features")
ax.spines[["top", "right"]].set_visible(False); plt.show()
```

**Result.** With stronger regularization CatBoost is the best single model on the holdout for the next-week and next-month **max** (≈ +5% skill), and on par with XGBoost for the next-week min. Like every model it does not beat the volatility band on the next-month min. *Caveat:* a first quick run had printed holdout numbers before this setting was chosen, so the CatBoost holdout result is not 100% blind.
